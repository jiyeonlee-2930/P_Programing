```python
##2. 3의 배수의 갯수 세가
count = 0
arr = [2,3,6,9,12,15,10,20,22,25]
arr_len = 10
for i in range(arr_len):
    if arr[i] % 3 == 0:
        count += 1
three = count
print(three)
```

```python
##5의 배수
count = 0
arr = [2,3,6,9,12,15,10,20,22,25]
arr_len = 10
for i in range(arr_len):
    if arr[i] % 5 == 0:
        count += 1
five = count
print(five)
```

```python
if three > five:
    print("three")
elif five > three:
    print("five")
else:
    print("smae")
```

```python
def func_a(arr, arr_len):   # 3의 배수 개수
    count = 0
    for i in range(arr_len):
        if arr[i] % 3 == 0:
            count += 1
    return count

def func_b(arr, arr_len):   # 5의 배수 개수
    count = 0
    for i in range(arr_len):
        if arr[i] % 5 == 0:
            count += 1
    return count

def func_c(three, five):    # 비교 후 문자열 반환
    if three > five:
        return "three"
    elif five > three:
        return "five"
    else:
        return "same"

def solution(arr, arr_len):
    three = func_a(arr, arr_len)
    five  = func_b(arr, arr_len)
    return func_c(three, five)

# 테스트
print(solution([2,3,6,9,12,15,10,20,22,25], 10))  # "three"
```

# 서로 다른 두 자연수 N 과 M 이 매개변수로 주어질 때 , N 부터 M 까지의 자연수 중에서 짝수들의 제곱의 합을 return 하도록 solution 함수 를 완성해 주세요.

```python
N = 4
M = 7
answer = 0
for i in range(N, M + 1):
    if i % 2 == 0: # 짝수인 경우
         answer += i * i
print(answer)
```

```python
def solution(N, M):
    answer = 0
    for i in range(N, M + 1):
        if i % 2 == 0:          # 짝수인 경우
            answer += i * i     # 제곱해서 더함
    return answer
```

# 3. 단어들이 들어 있는 배열에서 길이가 5 이상인 단어를 배열에 들어 있는 순서대로 이어 붙이려 합니다.

```python
words = ["my", "favorite", "color", "is", "violet"]
words_len = 5
answer = ""
for i in range(words_len):
    if len(words[i]) >= 5:      # 길이 5 이상인 단어만
        answer += words[i]      # 이어붙임
print(answer)
```

```python
def solution(words, words_len):
    answer = ""
    for i in range(words_len):
        if len(words[i]) >= 5:      # 길이 5 이상인 단어만
            answer += words[i]      # 이어붙임
    if answer == "":                # 없으면 "empty"
        return "empty"
    return answer
```

# 5. 게임 캐릭터가 몬스터와 1:1 전투를 하려 합니다 . 몬스터는 처음에 일정 수치의 체력을 가지고 있습니다 . 캐릭터가 전투에서 이기려면 몬스터를 공격해 몬스터의 체력을 0 이하로 만들어야 합니다 . 캐릭터는 공격 마법만 사용하며 , 공격 마법은 항상 같은 데미지를 입힙니다 . 몬스터는 힐링 마법만을 사용하며 , 힐링 마법은 항상 같은 수치로 체력을 회복합니다 . 둘은 항상 번갈아 가며 마법을 사용하고 ,처음에는 항상 캐릭터가 먼저 공격을 시작합니다.

##캐릭터의 공격 력 attack 과 몬스터가 자신의 차례에 회복하는 체력 recovery, 몬스터의 초기 체력 hp가 매개변수로 주어질 때 , 몬스터를 잡기 위해서 최소 몇 번 공격해야 하는지 return 하도록 solution 함수 를 작성하려 합니다 .

#1. attack 공격력, recovery 회복력, hp 초기체력

```python
attack = 30
recovery = 10
hp = 60
cnt = 0
while hp > 0:
    hp -= attack    # 캐릭터 공격
    cnt += 1      # 공격 횟수 증가
    if hp > 0:      # 몬스터가 살아있으면 회복
        hp += recovery
print(cnt)
```

```python
def solution(attack, recovery, hp):
    count = 0
    while hp > 0:
        hp -= attack    # 캐릭터 공격
        count += 1      # 공격 횟수 증가
        if hp > 0:      # 몬스터가 살아있으면 회복
            hp += recovery
    return count

# 테스트
print(solution(30, 10, 60))  # 3
```

# 6. 하루 동안 엘리베이터가 멈춘 층이 순서대로 들어 있는 배열이 있습니다 . 이때 , 엘리베이터의 총 이동 거리를 구하려 합니다 . 단 , 층과 층 사이의 거리는 1 입니다 .

```python
floors = [1, 2, 5, 4, 2]
floors_len = 5
answer = 0
for i in range(1, floors_len):
    answer += abs(floors[i] - floors[i-1])
print(answer)
```

```python
def solution(floors, floors_len):
    answer = 0
    for i in range(1, floors_len):
        answer += abs(floors[i] - floors[i-1])  # 절댓값으로 거리 계산
    return answer

# 테스트
print(solution([1,2,5,4,2], 5))  # 7
```

# 7. 화씨온도 를 섭씨온도 로 , 섭씨온도 를 화씨온도 로 바꾸려고 합니다 . 두 온도 사이의 환산 공식은 다음과 같습니다.
#환산 공식
# 화씨 온도 에서 섭씨 온도 로 환산 : 화씨 온도 32) ÷ 1.8 = 섭씨 온도
# 섭씨 온도 에서 화씨 온도 로 환산 : 섭씨 온도 x 1.8) + 32 = 화씨 온도

```python
def solution(value, unit):
    if unit == "F":                         # 화씨 → 섭씨
        answer = (value - 32) / 1.8
    else:                                   # 섭씨 → 화씨
        answer = (value * 1.8) + 32

    return int(answer)   # ← 수정된 줄: (float → int 로 정수 부분만 반환)

# 테스트
print(solution(527, "C"))  # 980
```

# 8. 자연수의 각 자릿수 중에서 소수는 몇 개인지 구하려 합니다 . 즉 , 자연수를 각 자릿수 별로 나누었을 때 , 2, 3, 5, 7 이 몇 개 있는지 구하려 합니다

```python
prime = [2, 3, 5, 7]
number = 29022531
digits = list(str(number))
print(digits)
answer = 0  # Initialize answer
for digit_str in digits:
    if int(digit_str) in prime:
        answer += 1
print(answer)
```

```python
def solution(number):
    prime = [2, 3, 5, 7]    # 한 자리 소수
    count = 0

    while number > 0:
        digit = number % 10         # 마지막 자릿수 추출
        if digit in prime:
            count += 1
        number = number // 10       # ← 수정된 줄: / 를 // 로 수정 (정수 나눗셈)
    return count

# 테스트
print(solution(29022531))  # 5
```

# 9. N명의 후보에 대해 투표한 결과가 들어 있는 배열이 있습니다 . 예를 들어 5 명의 후보에 대해 투표를 진행한 결과가 [2, 5, 3, 4, 1, 5, 1, 5, 5, 이라면 순서대로 [2 번 , 5 번 , 3 번 , 4 번 , 1 번 , 5 번 , 1 번 , 5 번 ,5 번 , 3 번 ] 후보에 투표했음을 나타냅니다 . 이때 , 정확히 K 표를 받은 후보는 총 몇 명인지 구하려 합니다 .

#

```python
votes =  [2, 5, 3, 4, 1, 5, 1, 5, 5, 3]
votes_len = 10
N = 5
K = 2
```

```python
def solution(votes, votes_len, N, K):
    count = [0] * (N + 1)       # 후보번호 1~N

    for i in range(votes_len):
        count[votes[i]] += 1    # 득표수 집계

    answer = 0
    for i in range(1, N + 1):   # ← 수정된 줄: range(1, N+1) 로 수정
        if count[i] == K:
            answer += 1
    return answer

# 테스트
print(solution([2,5,3,4,1,5,1,5,5,3], 10, 5, 2))  # 2
```

#10. A 백화점에서는 고객의 구매금액에 따라 다음과 같이 상품권을 지급합니다.
#100만원 -> 5만원상품권
#60만원 -> 3만원 상품권
#40만원 -> 2만원 상품권
#20만원 -> 1만원 상품권

```python
answer = 0
p = 0
purchase_len = 5
purchase = [150000, 210000, 399990, 990000, 1000000]
for i in range(purchase_len):
    p = purchase[i]
    if p >= 1000000:        # 100만원 이상
        answer += 50000
    elif p >= 600000:       # 60만원 이상
        answer += 30000
    elif p >= 400000:       # 40만원 이상
        answer += 20000
    elif p >= 200000:       # ← 수정된 줄: 20만원 이상
        answer += 10000
print(answer)
```

```python
def solution(purchase, purchase_len):
    answer = 0

    for i in range(purchase_len):
        p = purchase[i]
        if p >= 1000000:        # 100만원 이상
            answer += 50000
        elif p >= 600000:       # 60만원 이상
            answer += 30000
        elif p >= 400000:       # 40만원 이상
            answer += 20000
        elif p >= 200000:       # ← 수정된 줄: 20만원 이상
            answer += 10000
    return answer

# 테스트
print(solution([150000,210000,399990,990000,1000000], 5))  # 100000
```
