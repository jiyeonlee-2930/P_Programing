```python
##1. 반감기 10년, 최초 물질의 양 100g
## 최초 물질의 양이 1/10 이 될 때의 년도구하기
##반감기 : halfyear, 최초 물질의 양 : initial
##현재 물질의 양 current, 년도 : year
initial = 100
halfyear = 10
year = 0
current = initial
while current > initial /10.0:
    year += halfyear
    current -= current /2.0
    print(f"{year}년 {current}g")
print(f"{year}년 후에 잔류량은 {current}g 이다.")
```

```python
##2.종이 접기 면적 구하기
##종이를 접으면 면적이 1/2 일 때 1/100일 떄 접이 횟수
initial = 1
current = initial
cnt = 0
while 1:
    current /= 2
    cnt += 1
    print(f"{cnt} 회 : {current}")
    if current < initial / 100:
        break
    print(f"{cnt} 회 : {current}")
```

```python
##3. 종이를 접으면 높이가 2배 된다고 할 때
## 에베레스트 산의 높이 8800m 높아지려면 횟수 구하기
## 종이두께는 0.1 mm
initial = 0.1 ##mm
current = initial/1000 ##mm
cnt = 0
while 1:
    current *= 2
    cnt += 1
    print(f"{cnt} 회 : {current}m")
    if current > 8800:
        break
    print(f"{cnt} 회 : {current} m")
```

```python
##금액을 입금하면 잔액을 출력하는 함수
print("입금할 금액을 입력해주세요")
balance  =  500
input_money = int(input())
money = balance+input_money
print(f"잔액은 {money}원 입니다.")
```

```python
def deposit(balance):
    print("입금할 금액을 입력해주세요")
    input_money = int(input())
    money = balance+input_money
    print(f"잔액은 {money}원 입니다.")
    return money
```

```python
balance = deposit(500)
print(balance)
```

```python
balance = deposit(balance)
print(balance)
```

```python
def withdraw(balance):
    print("출금 금액 입력")
    money = int(input())
    if balance >= money:
        balance -= money
        print(f"{money}원을 출금하였습니다. 잔액은{balance}원 입니다.")
    return balance
```

```python
def DayCalculator(month,day):
    days = [31,28,31,30,31,60,31,31,30,31,30,31]
    len(days)
    t_day = 0
    for i in range(month-1):
        t_day += days[i]
        print(t_day)
        total_days = t_day + day-1
    print(total_days)
    return total_days
```

```python
month = 3
day = 3
DayCalculator(month,day)
```

```python
DayCalculator(2,2)
```

```python
solution = DayCalculator(3,3)-DayCalculator(2,2)
print(solution)
```
