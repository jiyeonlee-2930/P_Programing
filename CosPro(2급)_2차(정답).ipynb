# 1. 왼손 장갑의 제품 번호가 들어있는 리스트와 오른손 장갑의 제품 번호가 들어있는 리스트가 있습니다. 제품 번호는 1부터 10 사이의 자연수입니다. 제품 번호가 같은 왼손장갑과 오른손 장갑을 합쳐 장갑 한 쌍을 만들 수 있습니다. 이때, 최대한 많은 쌍의 장갑을 만들면 최대 몇 쌍을 만들 수 있는지 구하려 합니다.

1. 왼손 장갑이 제품 번호별로 몇 개씩 있는지 확인
2. 오른손 장갑이 제품 번호별로 몊 개씩 있는지 확인
3. 각 제품번호별로 최대한 많은 장갑 쌍을 만들며서 개수를 확인

```python
##제품 번호별로 장갑 갯수 구하기
number = 10
r_gloves = [1, 2, 2, 4, 4, 7]
l_gloves = [2,1,2,2,4]
r_counter = [0 for _ in range(number + 1)]
l_counter = [0 for _ in range(number + 1)]
for x in r_gloves:
    r_counter[x]+= 1
print(r_counter)
```

```python
for x in l_gloves:
    l_counter[x]+= 1
print(l_counter)
```

```python
##두 장갑의 제품번호를 비교하여 숫자가 작은 수만큼 장갑을 만들 수 있다.
total = 0
for i in range(number + 1):
    total += min(l_counter[i], r_counter[i])
print(total)
```

```python
###1.
max_product_number = 10
##
def func_a(gloves):
    counter = [0 for _ in range(max_product_number + 1)]
    for x in gloves:
        counter[x]+= 1
    return counter

def solution(left_gloves, right_gloves):
    left_counter = func_a(left_gloves)
    right_counter = func_a(right_gloves)

    total = 0
    for i in range(max_product_number + 1):
        total += min(left_counter[i], right_counter[i])
    return total

#아래는 테스트케이스 출력을 해보기 위한 코드입니다.
left_gloves = [2, 1, 2, 2, 4]
right_gloves = [1, 2, 2, 4, 4, 7]
ret = solution(left_gloves, right_gloves)

#[실행] 버튼을 누르면 출력 값을 볼 수 있습니다.
print("solution 함수의 반환 값은", ret, "입니다.")
```

# 2. 자연수가 들어있는 리스트에 3의 배수와 5의 배수 중 어떤수가 더 많은지 조사
1. 3의 배수의 갯수 조사
2. 5의 배수의 갯수 조사
3. 갯수를 비교하여 5의 배수가 많으면 five, 3의 배수가 많으면 3 이라고 출력

```python
arr = [2, 3, 6, 9, 12, 15, 10, 20, 22, 25]
#5의 배수의 갯수
count_5 = 0
for n in arr:
    if n % 5 == 0:
        count_5 += 1
print(count_5)
```

```python
count_3 = 0
for n in arr:
    if n % 3 == 0:
        count_3 += 1
print(count_3)
```

```python
if count_3 > count_5:
    print("three")
elif count_3 < count_5:
    print("five")
else:
    print("same")
```

```python
##2.
def func_a(arr): ## 5의 배수 갯수
    count = 0
    for n in arr:
        if n % 5 == 0:
            count += 1
    return count

def func_b(three, five): ## 3의배수와 5의 배수 숫자 비교
    if three > five:
        return "three"
    elif three < five:
        return "five"
    else:
        return "same"

def func_c(arr): ## 3의 배수 갯수
    count = 0
    for n in arr:
        if n % 3 == 0:
            count += 1
    return count

def solution(arr):
    count_three = func_c(arr)
    count_five = func_a(arr)
    answer = func_b(count_three, count_five)
    return answer

#아래는 테스트케이스 출력을 해보기 위한 코드입니다.
arr = [2, 3, 6, 9, 12, 15, 10, 20, 22, 25]
ret = solution(arr)

#[실행] 버튼을 누르면 출력 값을 볼 수 있습니다.
print("solution 함수의 반환 값은", ret, "입니다.")
```

## 5. 게임 캐릭터가 몬스터와 1:1 전투를 하려 합니다. 몬스터는 처음에 일정 수치의 체력(HP) 을 가지고 있습니다. 캐릭터가 전투에서 이기려면 몬스터를 공격해 몬스터의 체력을 0 이하로 만들어야합니다. 캐릭터는 공격 마법만 사용하며, 공격 마법은 항상 같은 데미지를 입힙니다. 몬스터는 힐링 마법만을 사용하며, 힐링 마법은 항상 같은 수치로 체력을 회복합니다. 둘은 항상 번갈아 가며 마법을 사용하고, 처음에는 항상 캐릭터가 먼저 공격을 시작
합니다.

1. 기본 체력을 갖고 있음
2. 캐릭터 : 공격마법사용
3. 몬스터 : 힐링마법 사용 체력 회복

```python
##초기 체력, 60, 공격 -30,힐링 +10
attack = 30
recovery = 10
hp = 60
count = 0
while(True):
    count += 1
    hp -= attack
    print(count, hp)
    if hp <= 0:
        break
    hp += recovery
    print(count, hp)
print(count)
```

```python
###5.
def solution(attack, recovery, hp):
    count = 0
    while(True):
        count += 1
        hp -= attack
        if hp <= 0:
            break
        hp += recovery
    return count

#아래는 테스트케이스 출력을 해보기 위한 코드입니다.
attack = 30
recovery = 10
hp = 60
ret = solution(attack, recovery, hp)

#[실행] 버튼을 누르면 출력 값을 볼 수 있습니다.
print("solution 함수의 반환 값은", ret, "입니다.")
```

# 6. 하루동안 이동한 엘리베이터의 총 이동거리를 구하라. 층과층 사이는 1이다.

```python
floors = [1, 2, 5, 4, 2]
dist = 0
length = len(floors)
for i in range(1,length):
    if floors[i] > floors[i-1]:
        dist +=  floors[i] - floors[i-1]
    else:
        dist += floors[i-1] - floors[i]
print(dist)
```

```python
##6.
def solution(floors):
    dist = 0
    length = len(floors)
    for i in range(1,length):
        if floors[i] > floors[i-1]:
            dist +=  floors[i] - floors[i-1]
        else:
            dist += floors[i-1] - floors[i]
    return dist

#아래는 테스트케이스 출력을 해보기 위한 코드입니다.
floors = [1, 2, 5, 4, 2]
ret = solution(floors)

#[실행] 버튼을 누르면 출력 값을 볼 수 있습니다.
print("solution 함수의 반환 값은", ret, "입니다.")
```

1. 예를 들어 K = 2일 때, 2표를 받은 후보는 1번, 3번, 총 2명이다.

```python
votes = [2, 5, 3, 4, 1, 5, 1, 5, 5, 3]
N = 5 ##후보자 수
K = 2 ##알고자 하는 득표수
counter = [0 for _ in range(N+1)]
for x in votes:
    counter[x] += 1
print(counter)
answer=0
for c in counter:
    if c == K:
        answer += 1
print(answer)
```

#10. A 백화점에서는 고객의 구매금액에 따라 다음과 같이 상품권을 지급합니다.
| 구매 금액 | 상품권 |
|-------------------|--------------|
| 100만 원 이상 구매 | 5만 원 상품권 |
| 60만 원 이상 구매 | 3만 원 상품권 |
| 40만 원 이상 구매 | 2만 원 상품권 |
| 20만 원 이상 구매 | 1만 원 상품권 |

```python
##고객의 구매 금액이 들어있는 리스트가 있을 때 상품권 총액을 구하여라.
purchase = [150000, 210000, 399990, 990000, 1000000]
total = 0
for p in purchase:
    if p >= 1000000:
        total += 50000
    elif p >= 600000:
        total += 30000
    elif p >= 400000:
        total += 20000
    elif p>= 200000:
        total += 10000
print(total)
```

```python
###10.
def solution(purchase):
    total = 0
    for p in purchase:
        if p >= 1000000:
            total += 50000
        elif p >= 600000:
            total += 30000
        elif p >= 400000:
            total += 20000
        elif p>= 200000:
            total += 10000
    return total

#아래는 테스트케이스 출력을 해보기 위한 코드입니다. 아래 코드는 잘못된 부분이 없으니, solution함수만 수정하세요.
purchase = [150000, 210000, 399990, 990000, 1000000]
ret = solution(purchase)


#[실행] 버튼을 누르면 출력 값을 볼 수 있습니다.
print("solution 함수의 반환 값은", ret, "입니다.")
```
