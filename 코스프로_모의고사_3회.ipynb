# 1. 학생들의 시험 점수가 주어졌을 때 n 번 학생이 몇 등인지 구하려 합니다. 학번은 0 번부터 시작하며,시험 점수는 학번 순으로 주어집니다.


#1. n 번 학생의 점수를 변수에 저장합니다
#2. 점수를 내림차순으로 정렬합니다.
#3. 배열의 첫 번째 원소부터 마지막 원소까지 순회하며 n 번 학생의 점수를 찾습니다.3-1. 1 번 단계에서 저장해 둔 점수와 같은 점수를 찾으면 등수를 return 합니다.



```python
scores = [20,60,98,59]
n = 2
target = scores[n-1]
sorted_scores = sorted(scores, reverse=True)
for rank, score in enumerate(sorted_scores, start=1):
    if score == target:
        print(rank)
        break
print(rank)
```

# 2. 모학교에서는 학기가 끝날 때마다 장학금을 줍니다 . 이때 장학생이 몇 명인지 구하려고 합니다 . 장학금을 주는 조건은 다음과 같습니다
#1.이번 학기 성적이 80 점 이상 (100 점 만점 이면서 석차가 상위 10% 이내인 학생)
#2.이번 학기 성적이 80 점 이상이면서 1 등인 학생
#3.직전 학기 대비 성적이 가장 많이 오른 학생 여러 명인 경우 해당 학생 전부

```python
current_grade = [70, 100, 70, 80, 50, 95]
current_grade_len = 6
last_grade = [35, 65, 80, 50, 20, 60]
last_grade_len = 6
```

```python
# 1. 학생별 석차 계산 (동점자는 같은 등수)
rank_list = []
for i in range(current_grade_len):
    r = 1
    for j in range(current_grade_len):
        if current_grade[j] > current_grade[i]:
            r += 1
    rank_list.append(r)
print(rank_list)
```

```python
# 2.학생별 성적 향상 최댓값 구하기고 해당하는 학생번호 출력
max_improvement = 0
for i in range(current_grade_len):
    diff = current_grade[i] - last_grade[i]
    if diff > max_improvement:
        max_improvement = diff
print(i, max_improvement)
```

```python
#3. 성적이 80점이고 석차가 상위 10% 이내인 학생 번호 출력
#석차가 10% => 0.6 이므로 해당하는 학생 없음.
top_10_percent = int(current_grade_len * 0.2)
print(top_10_percent)
for i in range(current_grade_len):
    if current_grade[i] >= 80 and rank_list[i] <= top_10_percent:
        print(i)
```

```python
# 4. 장학생 수 계산(중복수혜 제외)
top_10_percent = current_grade_len * 0.1
cnt = 0

for i in range(current_grade_len):
    diff = current_grade[i] - last_grade[i]

    # 세 조건을 or로 묶어서 하나라도 해당하면 1명만 카운트
    if (current_grade[i] >= 80 and rank_list[i] <= top_10_percent) or \
       (current_grade[i] >= 80 and rank_list[i] == 1) or \
       (diff == max_improvement and diff > 0):
        cnt += 1

print(f"장학생수 {cnt}명")
```

# 3. 체조선수는 여러 심사위원의 점수 중 가장 높은 점수 하나와 가장 낮은 점수 하나를 제외하고 나머지 점수들의 평균을 계산하여 최종 점수를 받습니다 . 단 , 이때 소수점 이하의 수는 버립니다

```python
scores = [35, 28, 98, 34, 20, 50, 85, 74, 71, 7]
scores_len = 10
##1. 최고 점수와 최저 점수 찾기
# 최고·최저 점수 찾기
max_score = scores[0]
for s in scores:
    if s > max_score:
        max_score = s
print(max_score)
```

```python
min_score = scores[0]
for s in scores:
    if s < min_score:
       min_score = s
print(min_score)
```

```python
##2. 최고와 최저 점수 제외하고 평균
total = 0
cnt = 0
avg = 0
for s in scores:
    if s == max_score or s == min_score:
        continue
    total += s
    cnt += 1
avg = total // cnt
print(f"평균 : {avg}")
```

# 4. 영어단어를 외우기 위해 단어를 반복하여 타이핑합니다 . 그런데 , 타이핑하고 보니 오타가 많습니다 . 오타를 수정하려면 문자를 몇 개 바꿔야 하는지 구하려 합니다

# 6. 타일을 'R', 'G', 'B' 색으로 칠하려 합니다 . R 색으로는 3 칸을 한 번에 , G 색으로는 2 칸을 한 번에 칠할 수 있으며 , B 색으로는 1 칸을 칠할 수 있습니다 . 색은 R, G, B 순서로 한 번씩 번갈아 가면서 사용해야 하며 , 타일의 길이를 넘겨서 칠할 수는 없습니다.

```python
tileLength = 16
one_set = "RRRGGB"   # 한 세트 = 6칸
set_size = 6
full_set = tileLength // set_size   # 완전한 세트 수
res = tileLength % set_size   # 남은 칸 수
print(full_set,res )
```

```python
# 나머지 칸을 순서대로 채울 수 있는지 확인
# one_set 앞에서부터 remainder 만큼 잘라서 붙이면 됨
if res == 1 or res == 2 or res == 4:
    result =  -1
else:
    result = one_set * full_set + one_set[:res]
print(result)
```

#7.  주스1 잔을 만들려면 사과 3 개와 당근 1 개가 필요합니다 . 그런데 키우는 토끼에게 먹이를 주기 위해 사과와 당근 종류에 상관없이 k 개를 빼놓으려고 합니다 . 주스는 최대한 많이 만들수록 좋습니다.

```python
##1. 주스 1잔 =  사과 3 + 당근 1개
apple  = 10
carrot = 5
k = 4
```

```python
answer = 0
## 사과는 당근의 3배가 되어야 한다.
##1. 사과의 수가 당근의 3배보다 작은 경우
if apple < (3 * carrot):
    answer = apple // 3
else:## 당근이 부족한 경우
    answer = carrot
##2.주스 만들고 남은 재료 계산
apple -= answer * 3
carrot -= answer
print(apple,carrot)
##3. 남은 재료로 k에서 차감
k = k - (apple + carrot)
print(k)
##4. k가 남아있으면, 주스를 다시 만들어야 함.
## 주스 1잔 = 사과3 + 당근 1 = 4, 4개 빼면 주스 1잔 감소
while k > 0:
    k -= 4
    answer -= 1
print(answer)
```

#9. 관공서 주차장에서는 차량 2 부제를 실시합니다 . 차량 2 부제는 차량 번호 끝자리가 홀수인 차량은 홀수 일에만 , 짝수인 차량은 짝수 일에만 주차장에 들어올 수 있도록 하는 제도입니다.

```python
day = 17
numbers = [3285, 1724, 4438, 2988, 3131, 2998]
numbers_len = 6
count = 0
for num in numbers:
    last_digit = num % 10   # 번호 끝자리
    # 홀수 날 & 홀수 번호
    if day % 2 == 1 and last_digit % 2 == 1:
        count += 1
    # 짝수 날 & 짝수 번호
    elif day % 2 == 0 and last_digit % 2 == 0:
        count += 1
print(count)
```
